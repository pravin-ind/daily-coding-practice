```python
name = "pravin"
tam_marks = 88
eng_marks = 94
mat_marks = 85
phy_marks = 74
che_marks = 68
com_marks = 98

print("name:", name)
print("tam_marks:", tam_marks)
print("eng_marks:", eng_marks)
print("mat_marks:", mat_marks)
print("phy_marks:", phy_marks)
print("che_marks:", che_marks)
print("com_marks:", com_marks)
print("total marks of", name, "=", (tam_marks+eng_marks+mat_marks+phy_marks+che_marks+com_marks),"/ 600")
print("average marks of",name, "=", (tam_marks+eng_marks+mat_marks+phy_marks+che_marks+com_marks)/6,"/ 100")
```
